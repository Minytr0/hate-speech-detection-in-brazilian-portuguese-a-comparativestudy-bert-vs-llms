```python
# Auxiliaries
import pandas as pd
import random
import time
import datetime
import numpy as np
import io
import re
```

```python
url = 'https://drive.google.com/file/d/15MTMse9MdIv-0-0yBtfkqks-5477e-R1/view?usp=sharing'
file_id = url.split('/')[-2]
read_url='https://drive.google.com/uc?id=' + file_id

# read the data
dataset = pd.read_csv(read_url, usecols=['text', 'hatespeech_comb'])

# display the first 5 rows
dataset.head()
```

```python
# 0 - hate speech
# 1 - offensive language

idenodio =  dataset.loc[dataset['hatespeech_comb'] == 0].sample(n = 50)
naoodio =  dataset.loc[dataset['hatespeech_comb'] == 1].sample(n = 50)

amostras =  list(idenodio['text'].values) + list(naoodio['text'].values)
classes = [0]*50 + [1]*50
```

```python
dic = {'text': amostras, 'hatespeech_comb': classes}
df = pd.DataFrame(dic)
df.to_csv('amostras.csv')
```
